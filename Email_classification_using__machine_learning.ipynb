```python
# Libraries import

import string

import numpy as np
import pandas as pd

import nltk

from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier


```

Downloading stopwords

```python
nltk.download("stopwords")
```

Load Dataset

```python
df = pd.read_csv("/content/sample_data/spam.csv")

df['label'] = (df['Category'] == 'spam').astype(int)
df.head(11)
```

Replacing line breaks with white space

```python
df["Message"] =df['Message'].apply(lambda x: x.replace('\r\n',''))
```

```python
df.Message.iloc[1] # getting message at position 1
```

```python
df.info()
```

```python
stemmer = PorterStemmer()

corpus =[]
stopwords_set = set(stopwords.words("english"))

for i in range(len(df)):
  text= df['Message'].iloc[i].lower()
  text = text.translate(str.maketrans('','', string.punctuation)).split()
  text= [stemmer.stem(word) for word in text if word not in stopwords_set]
  text= " ".join(text)
  corpus.append(text)
```

```python
vectorizer =CountVectorizer()

X = vectorizer.fit_transform(corpus).toarray()
y= df.label

X_train , X_test, y_train ,y_test = train_test_split(X,y ,test_size=0.2)
```

```python
clf = RandomForestClassifier(n_jobs = -1)

clf.fit(X_train,y_train)
```

Accuracy score for our model


```python
clf.score(X_test, y_test)
```
